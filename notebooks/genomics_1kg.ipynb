# 1000 Genomes-style SPCA workflow (synthetic smoke)

This notebook executes a tiny synthetic-genotype smoke, not a 1000 Genomes result. Replace `genotype_path` and `metadata_path` with verified prepared inputs for an actual analysis. The canonical matrix orientation is samples by SNPs. There is no fallback solver. Each raw bootstrap refits preprocessing, and population prediction uses one deterministic held-out split. Optional SPCA alpha grids are selected on a nested training-only validation split; the smoke uses the supported PCA default. Downloading and PLINK preprocessing remain an explicit external data-preparation step.

```python
from pathlib import Path
ROOT = Path.cwd().resolve().parent if Path.cwd().name == 'notebooks' else Path.cwd().resolve()
ROOT
```

```python
output_dir = ROOT / 'artifacts' / 'genomics_smoke'
input_dir = output_dir / 'input'
generator = ROOT / 'tools' / 'generate_genomics_smoke_data.py'
%run "$generator" --output-dir "$input_dir"
genotype_path = input_dir / 'synthetic_genotypes.npz'
metadata_path = input_dir / 'synthetic_populations.csv'
genotype_path, metadata_path
```

```python
script = ROOT / 'experiments' / 'genomics_1kg.py'
output_json = output_dir / 'results.json'
output_csv = output_dir / 'metrics.csv'
%run "$script" "$genotype_path" --orientation samples-by-snps --missing-value -9 --method pca --rank 2 --bootstrap-repeats 1 --metadata "$metadata_path" --output-json "$output_json" --output-csv "$output_csv"
```
